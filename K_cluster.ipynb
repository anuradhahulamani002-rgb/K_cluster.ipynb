### K means cluster 

### Identifies K cluster (grouping) within the dataset

### Steps :

> 1: Choose 3 (based on K value) random points to be centroids (mean)

> 2: Calculate distance between points and centroids assigning points to closest centroid 

> 3: Compute new centroids

Note: We need to repeat step 2 and step 3 until we reach stability (centroids value gets repeated)

```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import load_iris
iris=load_iris()
```

```python
iris
```

```python
iris.data
```

```python
iris.feature_names
```

```python
columns=['sepal length (cm)',
 'sepal width (cm)',
 'petal length (cm)',
 'petal width (cm)']
```

```python
columns = iris.feature_names
```

```python
columns
```

```python
df=pd.DataFrame(iris.data,columns=columns)
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
train ,test = train_test_split(df, test_size=0.2, random_state=42)
```

```python
train.shape
```

```python
test.shape
```

```python
from sklearn.cluster import KMeans

model = KMeans(n_clusters=3,random_state=42)
```

```python
model.fit(train)
```

```python
predict=model.predict(test)
```
